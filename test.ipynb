```python
import requests
import pandas as pd

OLLAMA_URL = "http://localhost:11434"

def get_installed_models():
    """Return list of installed model names."""
    try:
        r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10)
        r.raise_for_status()
        models = r.json().get("models", [])
        return [m["name"] for m in models]
    except Exception as e:
        print(f"Error fetching tags: {e}")
        return []

def get_model_details(model_name):
    """Fetch detailed info for one model."""
    try:
        r = requests.post(f"{OLLAMA_URL}/api/show", json={"name": model_name}, timeout=30)
        r.raise_for_status()
        data = r.json()
        info = data.get("model_info", {})
        return {
            "name": model_name,
            "parameter_size": info.get("general.parameter_count", "N/A"),
            "architecture": info.get("general.architecture", "N/A"),
            "context_length": info.get("llama.context_length", info.get("general.context_length", "N/A")),
            "quantization": info.get("general.quantization_version", "N/A"),
        }
    except Exception as e:
        return {"name": model_name, "parameter_size": "ERR", "architecture": "ERR",
                "context_length": "ERR", "quantization": "ERR", "error": str(e)}

def main():
    installed = get_installed_models()
    print(f"Found {len(installed)} installed models.\n")

    rows = []
    for name in installed:
        print(f"Fetching: {name}")
        rows.append(get_model_details(name))

    df = pd.DataFrame(rows)
    # Drop error column if it exists
    if "error" in df.columns:
        df = df.drop(columns=["error"])

    # Display
    pd.set_option("display.max_rows", None)
    pd.set_option("display.max_columns", None)
    pd.set_option("display.width", 200)
    print("\n=== Installed Model Details ===")
    print(df.to_string(index=False))

    # Optional: save to CSV
    df.to_csv("installed_model_details.csv", index=False)
    print("\nSaved to installed_model_details.csv")

if __name__ == "__main__":
    main()
```

```python
import requests
import json
import time
import csv
from datetime import datetime

OLLAMA_URL = "http://localhost:11434"
PROMPT = ("Explain the difference between a research hypothesis, "
          "a mathematical model, and a simulation model. "
          "Answer in 5 short bullet points.")

def get_installed_models():
    """Return list of installed model names."""
    try:
        r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10)
        r.raise_for_status()
        return [m["name"] for m in r.json().get("models", [])]
    except Exception as e:
        print(f"Error fetching models: {e}")
        return []

def test_model_streaming(model_name, prompt):
    """
    Test a single model with streaming enabled.
    Returns a dict with timing metrics and full response.
    """
    payload = {
        "model": model_name,
        "messages": [{"role": "user", "content": prompt}],
        "stream": True,
        "options": {"num_predict": 100}   # limit output tokens
    }
    result = {
        "model": model_name,
        "status": "ok",
        "ttft_ms": None,
        "total_time_s": None,
        "tokens_per_sec": None,
        "word_count": 0,
        "response": "",
        "error": "",
    }
    try:
        start = time.time()
        first_token_time = None
        word_times = []
        current_word = []
        response_text = ""

        with requests.post(f"{OLLAMA_URL}/api/chat", json=payload, stream=True, timeout=180) as resp:
            resp.raise_for_status()
            for line in resp.iter_lines(decode_unicode=True):
                if not line:
                    continue
                try:
                    data = json.loads(line)
                except json.JSONDecodeError:
                    continue

                if "message" in data and "content" in data["message"]:
                    token = data["message"]["content"]
                    if token:
                        now = time.time()
                        if first_token_time is None:
                            first_token_time = now
                            ttft = now - start
                            result["ttft_ms"] = round(ttft * 1000, 1)

                        # record word-level timing: accumulate characters until whitespace
                        for ch in token:
                            if ch in (" ", "\n", "\t", ".", ",", "!", "?"):
                                if current_word:
                                    word = "".join(current_word)
                                    word_times.append((word, now - start))
                                    current_word = []
                            else:
                                current_word.append(ch)
                        response_text += token

                if data.get("done", False):
                    break

        # flush remaining word
        if current_word:
            word_times.append(("".join(current_word), time.time() - start))

        end = time.time()
        result["total_time_s"] = round(end - start, 2)
        result["word_count"] = len(response_text.split())
        result["response"] = response_text.strip()
        if first_token_time:
            result["tokens_per_sec"] = round(len(response_text.split()) / (end - first_token_time), 2)

        # optional: save detailed word times to separate CSV for one model
        # for now, we just record count and maybe average gap
        if len(word_times) > 1:
            gaps = [t2 - t1 for (_, t1), (_, t2) in zip(word_times, word_times[1:])]
            result["avg_word_gap_ms"] = round(sum(gaps)/len(gaps)*1000, 1)
        else:
            result["avg_word_gap_ms"] = 0

    except Exception as e:
        result["status"] = "error"
        result["error"] = str(e)

    return result

def main():
    models = get_installed_models()
    print(f"Found {len(models)} installed models.\n")

    all_results = []
    for i, model in enumerate(models, 1):
        print(f"[{i}/{len(models)}] Testing {model}...")
        res = test_model_streaming(model, PROMPT)
        all_results.append(res)
        # Print quick summary
        if res["status"] == "ok":
            print(f"  TTFT: {res['ttft_ms']} ms, Total: {res['total_time_s']} s, "
                  f"Tokens/sec: {res['tokens_per_sec']}, Words: {res['word_count']}\n")
        else:
            print(f"  Error: {res['error']}\n")

    # Save to CSV
    csv_file = "model_test_results.csv"
    fieldnames = [
        "model", "status", "ttft_ms", "total_time_s",
        "tokens_per_sec", "word_count", "avg_word_gap_ms", "response", "error"
    ]
    with open(csv_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_results)
    print(f"\nResults saved to {csv_file}")

if __name__ == "__main__":
    main()
```

```python
import requests
import json
import time
import csv

OLLAMA_URL = "http://localhost:11434"
PROMPT = ("Explain the difference between a research hypothesis, "
          "a mathematical model, and a simulation model. "
          "Answer in 5 short bullet points.")

SKIP_KEYWORDS = [
    "cloud",        # cloud placeholders
    "base",         # base models (not instruction tuned)
    "shield",       # safety classifiers
    "codegemma:2b-code",  # code completion
]

def get_installed_models():
    try:
        r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10)
        r.raise_for_status()
        return [m["name"] for m in r.json().get("models", [])]
    except Exception as e:
        print(f"Error fetching models: {e}")
        return []

def is_chat_capable(name):
    """Rough filter to skip obvious non-chat models."""
    name_lower = name.lower()
    for kw in SKIP_KEYWORDS:
        if kw in name_lower:
            return False
    # Vision models might still work, but many need images. Skip them for text-only.
    if "vl" in name_lower or "vision" in name_lower:
        return False
    return True

def chat(model_name, prompt, stream=False, timeout=300):
    payload = {
        "model": model_name,
        "messages": [{"role": "user", "content": prompt}],
        "stream": stream,
        "options": {"num_predict": 100}
    }
    if not stream:
        # Non-streaming used for warm-up
        r = requests.post(f"{OLLAMA_URL}/api/chat", json=payload, timeout=timeout)
        r.raise_for_status()
        return r.json().get("message", {}).get("content", "")
    else:
        # Streaming used for measurement
        start = time.time()
        first_token_time = None
        response_text = ""
        with requests.post(f"{OLLAMA_URL}/api/chat", json=payload, stream=True, timeout=timeout) as resp:
            resp.raise_for_status()
            for line in resp.iter_lines(decode_unicode=True):
                if not line:
                    continue
                data = json.loads(line)
                if "message" in data and "content" in data["message"]:
                    token = data["message"]["content"]
                    if token:
                        if first_token_time is None:
                            first_token_time = time.time()
                            ttft = first_token_time - start
                        response_text += token
                if data.get("done"):
                    break
        total_time = time.time() - start
        return {
            "ttft_ms": round((first_token_time - start) * 1000, 1) if first_token_time else None,
            "total_time_s": round(total_time, 2),
            "tokens_per_sec": round(len(response_text.split()) / (total_time - ttft) if first_token_time and total_time > ttft else 0, 2),
            "word_count": len(response_text.split()),
            "response": response_text.strip()
        }

def main():
    models = get_installed_models()
    print(f"Found {len(models)} installed models.\n")

    results = []
    for i, name in enumerate(models, 1):
        if not is_chat_capable(name):
            print(f"[{i}/{len(models)}] Skipping {name} (not chat-capable)")
            results.append({"model": name, "status": "skipped"})
            continue

        print(f"[{i}/{len(models)}] Warming up {name}...")
        # Warm-up call (non-streaming) to load model
        try:
            chat(name, "Hello", stream=False, timeout=300)
            time.sleep(0.5)  # small pause
        except Exception:
            pass  # ignore warm-up errors

        print(f"  Measuring {name}...")
        try:
            res = chat(name, PROMPT, stream=True, timeout=300)
            res["model"] = name
            res["status"] = "ok"
            results.append(res)
            print(f"  TTFT: {res['ttft_ms']} ms, Total: {res['total_time_s']} s, "
                  f"Tokens/sec: {res['tokens_per_sec']}, Words: {res['word_count']}\n")
        except Exception as e:
            print(f"  Error: {e}\n")
            results.append({"model": name, "status": "error", "error": str(e)})

    # Save results
    csv_file = "model_test_results_improved.csv"
    fieldnames = ["model", "status", "ttft_ms", "total_time_s", "tokens_per_sec", "word_count", "response", "error"]
    with open(csv_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        for res in results:
            # Ensure all keys exist
            row = {k: res.get(k, "") for k in fieldnames}
            writer.writerow(row)
    print(f"\nResults saved to {csv_file}")

if __name__ == "__main__":
    main()
```

```python
import requests
import json
import base64
import time
import os
from IPython.display import display, Markdown, clear_output

# ---------- Image helpers (unchanged) ----------
def image_to_base64_from_path(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def image_to_base64_from_url(image_url):
    resp = requests.get(image_url, timeout=30)
    resp.raise_for_status()
    return base64.b64encode(resp.content).decode("utf-8")

def image_to_base64(image_input):
    if image_input.startswith(("http://", "https://")):
        return image_to_base64_from_url(image_input)
    else:
        return image_to_base64_from_path(image_input)

# ---------- Notebook markdown streaming version ----------
def chat_with_thinking_markdown(prompt, model="qwen3:4b", image_input=None):
    """
    Streams response from a thinking model and updates a markdown cell.
    """
    message = {"role": "user", "content": prompt}
    if image_input:
        try:
            message["images"] = [image_to_base64(image_input)]
        except Exception as e:
            print(f"Error loading image: {e}")
            return

    payload = {
        "model": model,
        "messages": [message],
        "stream": True,
    }

    resp = requests.post(
        "http://localhost:11434/api/chat",
        json=payload,
        stream=True,
        timeout=300,
    )
    resp.raise_for_status()

    # Timers
    start_time = time.time()
    first_thinking_time = None
    first_content_time = None
    thinking_text = ""
    content_text = ""

    # Create a display handle to update markdown
    handle = display(Markdown(""), display_id=True)

    for line in resp.iter_lines(decode_unicode=True):
        if not line:
            continue
        data = json.loads(line)
        msg = data.get("message", {})

        # Thinking
        if "thinking" in msg and msg["thinking"]:
            thinking_text += msg["thinking"]
            if first_thinking_time is None:
                first_thinking_time = time.time()

        # Content
        if "content" in msg and msg["content"]:
            content_text += msg["content"]
            if first_content_time is None:
                first_content_time = time.time()

        # Build markdown with current text
        md = f"## 🤔 Thinking\n\n{thinking_text}\n\n---\n\n## 💬 Answer\n\n{content_text}"
        handle.update(Markdown(md))

        if data.get("done"):
            break

    total_time = time.time() - start_time

    # Final update with timings
    timings = f"\n\n---\n**TTFT (thinking):** {first_thinking_time - start_time:.3f}s  \n" if first_thinking_time else ""
    timings += f"**TTFT (content):** {first_content_time - start_time:.3f}s  \n" if first_content_time else ""
    timings += f"**Total time:** {total_time:.3f}s"

    handle.update(Markdown(md + timings))

    return thinking_text, content_text
```

```python
!pip install ipywidgets

import requests
import json
import base64
import time
import os
from pathlib import Path
from IPython.display import display, Markdown, clear_output
from ipywidgets import interact, widgets, FileUpload

# ---------- Base helpers ----------
def image_to_base64_from_path(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def image_to_base64_from_url(image_url):
    resp = requests.get(image_url, timeout=30)
    resp.raise_for_status()
    return base64.b64encode(resp.content).decode("utf-8")

def image_to_base64(image_input):
    """image_input can be path, URL, or bytes."""
    if isinstance(image_input, bytes):
        return base64.b64encode(image_input).decode("utf-8")
    elif isinstance(image_input, str) and image_input.startswith(("http://", "https://")):
        return image_to_base64_from_url(image_input)
    else:
        return image_to_base64_from_path(image_input)

# ---------- Detect model capabilities ----------
def get_model_capabilities(model):
    """
    Returns dict:
    {
      "thinking": bool,
      "vision": bool,
      "architecture": str,
      "params": str
    }
    Detection order:
      1. Ollama /api/show (most accurate)
      2. Model name heuristics
    """
    caps = {
        "thinking": False,
        "vision": False,
        "architecture": "unknown",
        "params": "unknown",
    }

    # 1. Try Ollama /api/show
    try:
        r = requests.post("http://localhost:11434/api/show", json={"name": model}, timeout=10)
        if r.status_code == 200:
            data = r.json()
            info = data.get("model_info", {})
            caps["architecture"] = info.get("general.architecture", "unknown")
            caps["params"] = info.get("general.parameter_count", "unknown")

            # Check capabilities from model_info
            # Vision often indicated by projectors or vision tower
            arch_str = json.dumps(info).lower()
            if "vision" in arch_str or "clip" in arch_str or "projector" in arch_str:
                caps["vision"] = True

            # Thinking is not always in model_info, but check common keys
            if "thinking" in arch_str or "reasoning" in arch_str:
                caps["thinking"] = True
    except Exception:
        pass

    # 2. Name heuristics (fallback)
    m = model.lower()
    if any(k in m for k in ["thinking", "reasoning", "r1"]):
        caps["thinking"] = True
    if any(k in m for k in ["vl", "vision", "vlm", "multimodal", "llava"]):
        caps["vision"] = True

    return caps

# ---------- Main streaming chat ----------
def chat_with_auto_markdown(prompt,
                            model="qwen3:4b",
                            image_input=None,
                            show_features=True):
    """
    Streams response from any Ollama model.
    Automatically detects thinking / vision.
    Updates markdown output cell.
    """
    # Detect features
    caps = get_model_capabilities(model)

    # Build message
    message = {"role": "user", "content": prompt}

    if image_input is not None:
        try:
            image_b64 = image_to_base64(image_input)
            message["images"] = [image_b64]
            caps["vision"] = True   # assume image supplied
        except Exception as e:
            print(f"Error loading image: {e}")
            return None, None

    # If no image but model is vision-only, warn
    if caps["vision"] and image_input is None:
        print("⚠️ Model supports vision, but no image provided. Text-only response may be poor.")

    payload = {
        "model": model,
        "messages": [message],
        "stream": True,
    }

    resp = requests.post(
        "http://localhost:11434/api/chat",
        json=payload,
        stream=True,
        timeout=300,
    )
    resp.raise_for_status()

    # Timers
    start_time = time.time()
    first_thinking_time = None
    first_content_time = None
    thinking_text = ""
    content_text = ""

    # Create handle
    handle = display(Markdown(""), display_id=True)

    # Header
    feature_str = f"""
**Model:** `{model}`  
**Thinking:** {'✅' if caps['thinking'] else '❌'}  
**Vision:** {'✅' if caps['vision'] else '❌'}  
**Architecture:** `{caps['architecture']}`  
**Params:** `{caps['params']}`  

---
"""
    handle.update(Markdown(feature_str))

    for line in resp.iter_lines(decode_unicode=True):
        if not line:
            continue
        data = json.loads(line)
        msg = data.get("message", {})

        # Thinking
        if "thinking" in msg and msg["thinking"]:
            thinking_text += msg["thinking"]
            if first_thinking_time is None:
                first_thinking_time = time.time()

        # Content
        if "content" in msg and msg["content"]:
            content_text += msg["content"]
            if first_content_time is None:
                first_content_time = time.time()

        # Build markdown
        md = feature_str

        if thinking_text:
            md += f"## 🤔 Thinking\n\n{thinking_text}\n\n---\n\n"

        md += f"## 💬 Answer\n\n{content_text}"

        handle.update(Markdown(md))

        if data.get("done"):
            break

    total_time = time.time() - start_time

    # Timings
    timing_md = "\n\n---\n"
    if first_thinking_time:
        timing_md += f"**TTFT (thinking):** {first_thinking_time - start_time:.3f}s  \n"
    if first_content_time:
        timing_md += f"**TTFT (content):** {first_content_time - start_time:.3f}s  \n"
    timing_md += f"**Total time:** {total_time:.3f}s"

    handle.update(Markdown(md + timing_md))

    return thinking_text, content_text
```

```python
import requests
import json
import base64
import time
import os
import pandas as pd
from IPython.display import display, Markdown

# ---------- 1. Model feature detection ----------
def get_model_features(model_name):
    """
    Determine if model has thinking/reasoning and vision.
    1. Try to load CSV and look up 'Input' column.
    2. If not found, use name heuristics.
    Returns (has_thinking, has_vision)
    """
    # Try to read combined CSV (if you have it)
    csv_path = r"E:\llm\Model\csv\all_models_combined.csv"
    input_type = None
    if os.path.exists(csv_path):
        try:
            df = pd.read_csv(csv_path)
            row = df[df['Name'] == model_name]
            if not row.empty:
                input_type = row.iloc[0].get('Input', '')
        except:
            pass

    # Vision detection from CSV if available
    if input_type is not None:
        has_vision = 'Image' in str(input_type)
    else:
        # Heuristic based on name
        low = model_name.lower()
        vision_patterns = ['vl', 'vision', 'llava', 'llama3.2-vision', 'llama4',
                           'gemma3', 'gemma4', 'ministral', 'mistral-small3.1',
                           'mistral-medium-3.5', 'medgemma', 'qwen2.5vl', 'qwen3-vl']
        has_vision = any(p in low for p in vision_patterns)

    # Thinking detection
    low = model_name.lower()
    thinking_patterns = ['thinking', 'reasoning', 'deepseek-r1', 'phi4-reasoning',
                         'phi4-mini-reasoning', 'nemotron-3.5-lightning',
                         'nemotron-3-nano', 'nemotron-3-super']
    has_thinking = any(p in low for p in thinking_patterns)
    # Some models may output reasoning without "thinking" in name (e.g., deepseek-r1)
    if 'deepseek-r1' in low:
        has_thinking = True

    return has_thinking, has_vision

# ---------- 2. Image helpers ----------
def image_to_base64_from_path(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def image_to_base64_from_url(image_url):
    resp = requests.get(image_url, timeout=30)
    resp.raise_for_status()
    return base64.b64encode(resp.content).decode("utf-8")

def image_to_base64(image_input):
    if image_input.startswith(("http://", "https://")):
        return image_to_base64_from_url(image_input)
    else:
        return image_to_base64_from_path(image_input)

# ---------- 3. Main chat function ----------
def chat_aware(prompt, model="qwen3:4b", image_input=None):
    """
    Chat with automatic handling of thinking/reasoning and vision.
    - If model supports vision and no image_input given, ask user for path/URL.
    - Displays model features in the output.
    """
    has_thinking, has_vision = get_model_features(model)

    # If vision model and no image provided, ask interactively
    if has_vision and image_input is None:
        print(f"🔍 Model '{model}' supports images.")
        img = input("Enter image path or URL (or press Enter to skip): ").strip()
        if img:
            image_input = img

    # Build message
    message = {"role": "user", "content": prompt}
    if image_input:
        try:
            image_b64 = image_to_base64(image_input)
            message["images"] = [image_b64]
        except Exception as e:
            print(f"❌ Error loading image: {e}")
            return

    payload = {
        "model": model,
        "messages": [message],
        "stream": True,
    }

    resp = requests.post(
        "http://localhost:11434/api/chat",
        json=payload,
        stream=True,
        timeout=300,
    )
    resp.raise_for_status()

    # Timers
    start_time = time.time()
    first_thinking_time = None
    first_content_time = None
    thinking_text = ""
    content_text = ""

    # Create display handle
    handle = display(Markdown(""), display_id=True)

    # Build feature header
    features = []
    if has_thinking:
        features.append("🧠 **Thinking**")
    if has_vision:
        features.append("👁️ **Vision**")
    features_str = ", ".join(features) if features else "None"
    header = f"### Model: {model}\n**Features:** {features_str}\n\n"

    for line in resp.iter_lines(decode_unicode=True):
        if not line:
            continue
        data = json.loads(line)
        msg = data.get("message", {})

        # Thinking
        if "thinking" in msg and msg["thinking"]:
            thinking_text += msg["thinking"]
            if first_thinking_time is None:
                first_thinking_time = time.time()

        # Content
        if "content" in msg and msg["content"]:
            content_text += msg["content"]
            if first_content_time is None:
                first_content_time = time.time()

        # Build markdown
        md = header
        if thinking_text:
            md += f"## 🤔 Thinking\n\n{thinking_text}\n\n---\n\n"
        md += f"## 💬 Answer\n\n{content_text}"

        handle.update(Markdown(md))

        if data.get("done"):
            break

    total_time = time.time() - start_time

    # Final timings
    timings = ""
    if first_thinking_time:
        timings += f"**TTFT (thinking):** {first_thinking_time - start_time:.3f}s  \n"
    if first_content_time:
        timings += f"**TTFT (content):** {first_content_time - start_time:.3f}s  \n"
    timings += f"**Total time:** {total_time:.3f}s"

    handle.update(Markdown(md + "\n---\n" + timings))

    return thinking_text, content_text
```

```python
# Call chat
prompt = "Explain the difference between a research hypothesis, a mathematical model, and a simulation model."
chat_aware(prompt, model="qwen2:0.5b")
```
