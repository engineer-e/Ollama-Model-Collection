```python
import pymupdf

FILE_NAME_PAPER = "2608.02980v1.pdf"
doc = pymupdf.open(FILE_NAME_PAPER)

colors = [
    (1, 0, 0, 0.3),       # Red
    (0, 1, 0, 0.3),       # Green
    (0, 0, 1, 0.3),       # Blue
    (1, 0.5, 0, 0.3),     # Orange
    (1, 0, 1, 0.3),       # Magenta
    (0, 1, 1, 0.3),       # Cyan
    (1, 1, 0, 0.3),       # Yellow
    (0.5, 0, 0.5, 0.3),   # Purple
    (0.1, 0.8, 0.1, 0.3), # Lime Green
    (0.8, 0.2, 0.8, 0.3), # Pink
]

print(f"Total pages: {doc.page_count}")

for pno in range(doc.page_count):
    page = doc[pno]
    links = page.get_links()
    print(f"Page {pno+1}: Found {len(links)} link(s).")
    
    if links:
        for idx, link in enumerate(links):
            # ------------- FIX HERE -------------
            rect = link["from"]   # use "from" instead of "rect"
            # ------------------------------------
            
            fill_color = colors[idx % len(colors)]
            page.draw_rect(
                rect,
                fill=fill_color,
                color=(0.1, 0.1, 0.1),
                width=0.5
            )
    
    mat = pymupdf.Matrix(1.5, 1.5)
    pix = page.get_pixmap(matrix=mat, alpha=False)
    pix.save(f"link_cover/page-{pno+1}-highlighted.png")
    print(f"  Saved: page-{pno+1}-highlighted.png")

doc.close()
print("\nAll pages processed successfully!")
```

```python
import pymupdf

FILE_NAME_PAPER = "2608.02980v1.pdf"
doc = pymupdf.open(FILE_NAME_PAPER)

# 10 distinct semi‑transparent colors (RGBA)
colors = [
    (1, 0, 0, 0.3),       # Red
    (0, 1, 0, 0.3),       # Green
    (0, 0, 1, 0.3),       # Blue
    (1, 0.5, 0, 0.3),     # Orange
    (1, 0, 1, 0.3),       # Magenta
    (0, 1, 1, 0.3),       # Cyan
    (1, 1, 0, 0.3),       # Yellow
    (0.5, 0, 0.5, 0.3),   # Purple
    (0.1, 0.8, 0.1, 0.3), # Lime
    (0.8, 0.2, 0.8, 0.3), # Pink
]

print(f"Total pages: {doc.page_count}")

for pno in range(doc.page_count):
    page = doc[pno]
    links = page.get_links()
    print(f"Page {pno+1}: {len(links)} link(s)")

    for idx, link in enumerate(links):
        # Correct key: "from" (the clickable rectangle)
        rect = link["from"]
        fill_color = colors[idx % len(colors)]
        page.draw_rect(
            rect,
            #fill=fill_color,
            color=(1, 0, 0),  # dark border
            width=1.5
        )

    # Render at 150 DPI (zoom = 150/72)
    mat = pymupdf.Matrix(150/72, 150/72)
    pix = page.get_pixmap(matrix=mat, alpha=False)
    pix.save(f"links/page-{pno+1:02d}-highlighted.png")
    print(f"  Saved page-{pno+1:02d}-highlighted.png")

doc.close()
print("\nAll done!")
```

# To Check Each Block

```python
import pymupdf

FILE_NAME_PAPER = "2608.02980v1.pdf"
doc = pymupdf.open(FILE_NAME_PAPER)

print(f"Total pages: {doc.page_count}")

for pno in range(doc.page_count):
    page = doc[pno]
    blocks = page.get_text("blocks")
    print(f"Page {pno+1}: {len(blocks)} block(s)")

    for block in blocks:
        # Tuple: (x0, y0, x1, y1, text, block_no, block_type)
        rect = pymupdf.Rect(block[0], block[1], block[2], block[3])

        # ✅ Draw ONLY a border – text stays perfectly visible
        page.draw_rect(rect, color=(1, 0, 0), width=1.5)   # Red border, 1.5px thick

    # Render the page
    mat = pymupdf.Matrix(150/72, 150/72)  # 150 DPI
    pix = page.get_pixmap(matrix=mat, alpha=False)
    pix.save(f"border/page-{pno+1:02d}-blocks-border.png")
    print(f"  Saved: page-{pno+1:02d}-blocks-border.png")

doc.close()
print("\n✅ Done! Text is fully readable, boundaries clearly visible.")
```

# Extract Text

```python
import pymupdf

FILE_NAME_PAPER = "2608.02980v1.pdf"
doc = pymupdf.open(FILE_NAME_PAPER)
out = open("output.txt", "wb") # create a text output
for page in doc: # iterate the document pages
    text = page.get_text().encode("utf8") # get plain text (is in UTF-8)
    out.write(text) # write text of page
    out.write(bytes((12,))) # write page delimiter (form feed 0x0C)
out.close()
doc.close()
```

# Extract Image


```python
import pymupdf

FILE_NAME_PAPER = "2608.02980v1.pdf"
doc = pymupdf.open(FILE_NAME_PAPER)


for page_index in range(len(doc)): # iterate over pdf pages
    page = doc[page_index] # get the page
    image_list = page.get_images()

    # print the number of images found on the page
    if image_list:
        print(f"Found {len(image_list)} images on page {page_index}")
    else:
        print("No images found on page", page_index)

    for image_index, img in enumerate(image_list, start=1): # enumerate the image list
        xref = img[0] # get the XREF of the image
        pix = pymupdf.Pixmap(doc, xref) # create a Pixmap

        if pix.n - pix.alpha > 3: # CMYK: convert to RGB first
            pix = pymupdf.Pixmap(pymupdf.csRGB, pix)

        pix.save(f"image/page_{page_index}-image_{image_index}.png") # save the image as png
        pix = None
```
