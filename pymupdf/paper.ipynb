# Abstract

- Descriptive 
  - 100-200 word
  - Indicates type of information found in the paper
  - It explain - purpose, objective, methods , Not (results & conclusion)

- Informative 
  - Most common for conference and journal papers
  - "Surrogate" of the research paper
  - summarize ALL parts of the paper
  - Know and Follow the guidelines
  - Max/Min Word count
  - Structured or Unstructed
  - Check formatting requirements


- 5 section
  - Purpose and Motivation
    - Why this Particular study?
    - Why is it important? area of study?
  - Problem
    - What is your research trying to understand or solve?
    - What is your central claim or argument?
    - "The Purpose of this research is ..."
  - Method
    - Discuss only the most significant methods used
  - Results
    - State only the most significant results
    - Relate with problem and conclusion
      - Trends
      - Figures
      - Correlations
      - Any interesting outcomes?  
  - Conclusion 
    - Relate with problem and purpose
    - What do the results mean in the context of the problem?
    - What other unanswered questions are there?

- Tip

1. Write the Abstract Last
2. Keep your content in correct order
3. Write the abstract from scratch
4. Only use highly relevant terms
5. Relevant terms = Pertinent search results

This video, led by *Professor David Stuckler*, provides a structured, 15-minute workflow for writing a winning academic abstract. Rather than agonizing over the abstract at the start, *Professor Stuckler* recommends writing it **after** the discussion section but **before** finalizing the introduction (2:13).

**Core Abstract Strategy:**
* **The Mental Model:** Treat your abstract as a "highlight reel" or a mini-snapshot of your entire paper. It must be concise, show study credibility, and be internally coherent with your results (1:13).
* **The 4-Part Structure:** Regardless of whether your journal requires a structured or unstructured abstract, use this template to ensure all key elements are present:
    1. **Background (6:52):** Use 2–3 sentences to establish the strategic importance of your study and identify the "gap" in existing literature.
    2. **Methods (8:27):** Be linear and brief. Focus on what you did (data source, analysis) rather than justifying why you did it.
    3. **Results (9:30):** Present your 3–5 main findings directly, showing rather than telling. Use evidence to bolster credibility.
    4. **Conclusions (10:43):** Offer 1–2 key takeaways and potential implications for policy or future research.

**Common Pitfalls:**
Avoid including citations, tables, figures, excessive jargon, or overly detailed methodological justifications (11:52). The goal is to provide enough clarity so that editors and reviewers can quickly grasp the value and rigor of your work.

- Mental Model For Abstracts
  - A Mini Snapshot of your paper
  - Concise Writing is hard
  - Showcase your main finding
  - Credibility needs to emerge quickly
  - Readers should see it at a glance
  - Aligns with your paper
  - If the abstract is unclear, the paper itself is unclear
  - Powerful Diagnostic tool - yourself, reviewers/editors
  
- When to Write Your Abstracts
  - Method -> Results -> Discussion -> Abstract -> Introduction
  - Once the abstract is clear, the introduction writes itself you know exactly what to justify,
  - Write too early and you guessing
